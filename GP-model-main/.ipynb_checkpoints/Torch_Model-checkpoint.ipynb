```python
import os
import torch
from torchaudio.transforms import Resample
import torchaudio
import whisper 
import nemo
import numpy as np
import soundfile as sf
import librosa
transform = Resample(orig_freq=48000, new_freq=16000)
```

```python
model, decoder, utils = torch.hub.load(repo_or_dir='snakers4/silero-models',
                                               model='silero_stt',
                                               language='en',
                                               device=torch.device('cpu'))
```

```python
class NeuralASR:
    word_locations_in_samples = None
    audio_transcript = None

    def __init__(self, model: torch.nn.Module, decoder) -> None:
        super().__init__()
        self.model = model
        self.decoder = decoder  # Decoder from CTC-outputs to transcripts

    def getTranscript(self) -> str:
        """Get the transcripts of the process audio"""
        assert(self.audio_transcript != None,
               'Can get audio transcripts without having processed the audio')
        return self.audio_transcript

    def getWordLocations(self) -> list:
        """Get the pair of words location from audio"""
        assert(self.word_locations_in_samples != None,
               'Can get word locations without having processed the audio')

        return self.word_locations_in_samples

    def processAudio(self, audio: torch.Tensor):
        """Process the audio"""
        audio_length_in_samples = audio.shape[1]
        with torch.inference_mode():
            nn_output = self.model(audio)

            self.audio_transcript, self.word_locations_in_samples = self.decoder(
                nn_output[0, :, :].detach(), audio_length_in_samples, word_align=True)
```

```python
asrModel = NeuralASR(model, decoder)
```

```python
def asr(path_audio):
    waveform, sample_rate = torchaudio.load(path_audio, normalize=True)
    waveform = transform(waveform)
    mat = torchaudio.info(path_audio)

    print(mat)

    asrModel.processAudio(waveform)
    return asrModel.getTranscript()
```

```python
wh_model = whisper.load_model('large-v3')
```

```python
def transcribe(path_audio, wh_model = wh_model):
    waveform, sample_rate = torchaudio.load(path_audio)
    waveform = transform(waveform)
    waveform = waveform.numpy()

    audio = np.squeeze(waveform)

    audio = whisper.pad_or_trim(audio)

    mel = whisper.log_mel_spectrogram(audio).to(wh_model.device)
    _, probs = wh_model.detect_language(mel)
    print(f"language is : {max(probs, key=probs.get)}")
    
    res = wh_model.transcribe(audio, fp16=False)
    
    return res['text']
```

```python
path_audio = './night.wav'
```

```python
transcribe(path_audio)
```

```python
asr(path_audio)
```

## Wav2Vec 960h

```python
from transformers import Wav2Vec2Config, Wav2Vec2Model

config = Wav2Vec2Config()

model = Wav2Vec2Model(config)

config = model.config
```

```python
def load_wav2vec_960h():
    from transformers import Wav2Vec2ForCTC, Wav2Vec2Tokenizer
    # import model, feature extractor, tokenizer
    model = Wav2Vec2ForCTC.from_pretrained("facebook/wav2vec2-base-960h")
    tokenizer = Wav2Vec2Tokenizer.from_pretrained("facebook/wav2vec2-base-960h")
    return model, tokenizer

def load_hubert():
    from transformers import Wav2Vec2Processor, HubertForCTC

    processor = Wav2Vec2Processor.from_pretrained("facebook/hubert-large-ls960-ft")
    model = HubertForCTC.from_pretrained("facebook/hubert-large-ls960-ft")
    return model, processor

def transcription_wav2vec(tokenizer, model, audio_file):
    speech, sr = sf.read(audio_file)
    
    if len(speech.shape)> 1:
        speech = speech[:,0] + speech[:, 1]
    
    if sr != 16000:
        speech = librosa.resample(y=speech, orig_sr=sr, target_sr=16000)
    input_values = tokenizer(speech, return_tensors="pt").input_values
    logits = model(input_values).logits
    
    pred_ids = torch.argmax(logits, axis=-1)
    
    transcribe = tokenizer.decode(pred_ids[0])
    return transcribe
```

```python
model, tokenizer = load_wav2vec_960h()
txt = transcription_wav2vec(tokenizer,model, path_audio)
txt
```
